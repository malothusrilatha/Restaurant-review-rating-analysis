```python
# Convert ratings to numeric values
df_clean['rate'] = (
    df_clean['rate']
    .astype(str)
    .str.replace('/5', '', regex=False)
    .str.strip()
)

df_clean['rate'] = pd.to_numeric(
    df_clean['rate'], errors='coerce'
)

# Remove rows without valid ratings
df_clean = df_clean.dropna(subset=['rate'])

# Display top 10 restaurants by rating
top_restaurants = df_clean.sort_values(
    by='rate', ascending=False
)[['name', 'rate', 'votes']].head(10)

display(top_restaurants)
```

```python
import matplotlib.pyplot as plt

# Select top 10 restaurants by rating
top_restaurants = (
    df_clean.sort_values(by='rate', ascending=False)
    [['name', 'rate']]
    .head(10)
)

# Create bar chart
plt.figure(figsize=(12, 6))

plt.bar(
    top_restaurants['name'],
    top_restaurants['rate']
)

plt.title('Top 10 Restaurants by Rating')
plt.xlabel('Restaurant Name')
plt.ylabel('Rating')
plt.xticks(rotation=45, ha='right')
plt.ylim(0, 5)
plt.tight_layout()
plt.show()
```

```python

import pandas as pd
import matplotlib.pyplot as plt

# Convert votes to numeric values
df_clean['votes'] = pd.to_numeric(
    df_clean['votes'], errors='coerce'
)

# Group votes by restaurant name
top_votes = (
    df_clean.groupby('name', as_index=False)['votes']
    .sum()
    .sort_values(by='votes', ascending=False)
    .head(10)
)

print("Top 10 Restaurants by Total Votes:")
display(top_votes)

# Create bar chart
plt.figure(figsize=(12, 6))
plt.bar(top_votes['name'], top_votes['votes'])
plt.title('Top 10 Restaurants by Total Votes')
plt.xlabel('Restaurant Name')
plt.ylabel('Total Votes')
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()
```

```python


import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Create a copy of the cleaned dataset
heatmap_data = df_clean.copy()

# Convert columns to numeric safely
heatmap_data['rate'] = pd.to_numeric(
    heatmap_data['rate'], errors='coerce'
)

heatmap_data['votes'] = pd.to_numeric(
    heatmap_data['votes'], errors='coerce'
)

# Clean cost values
cost_col = 'approx_cost(for two)'

heatmap_data[cost_col] = pd.to_numeric(
    heatmap_data[cost_col]
    .astype(str)
    .str.replace(',', '', regex=False)
    .str.strip(),
    errors='coerce'
)

# Select numeric columns and calculate correlation
numeric_data = heatmap_data[
    ['rate', 'votes', cost_col]
].dropna()

correlation = numeric_data.corr()

# Draw heatmap
plt.figure(figsize=(8, 5))
sns.heatmap(
    correlation,
    annot=True,
    cmap='coolwarm',
    fmt='.2f'
)

plt.title('Correlation Between Rating, Votes and Cost')
plt.tight_layout()
plt.show()

```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Compare ratings based on online order availability
order_rating = df_clean.groupby('online_order')['rate'].mean()

print("Average Rating Based on Online Orders:")
print(order_rating)

# Draw bar chart
plt.figure(figsize=(7, 5))
sns.barplot(
    data=df_clean,
    x='online_order',
    y='rate',
    errorbar=None
)

plt.title('Online Order vs Average Restaurant Rating')
plt.xlabel('Online Order Available')
plt.ylabel('Average Rating')
plt.show()
```

```python
 import pandas as pd
import matplotlib.pyplot as plt

# Find the cost column automatically
cost_col = next(
    (col for col in df_clean.columns
     if 'approx_cost' in col.lower()),
    None
)

if cost_col is None:
    print("Cost column not found!")
    print("Available columns:", df_clean.columns.tolist())

else:
    # Clean the cost values
    df_clean[cost_col] = pd.to_numeric(
        df_clean[cost_col]
        .astype(str)
        .str.replace(',', '', regex=False)
        .str.strip(),
        errors='coerce'
    )

    # Calculate cost details
    average_cost = df_clean[cost_col].mean()
    lowest_cost = df_clean[cost_col].min()
    highest_cost = df_clean[cost_col].max()

    print("Restaurant Cost Analysis")
    print("------------------------")
    print("Cost Column:", cost_col)
    print("Average Cost for Two: ₹", round(average_cost, 2))
    print("Lowest Cost: ₹", lowest_cost)
    print("Highest Cost: ₹", highest_cost)

    # Plot cost distribution
    plt.figure(figsize=(8, 5))
    plt.hist(
        df_clean[cost_col].dropna(),
        bins=20,
        edgecolor='black'
    )

    plt.title('Distribution of Restaurant Cost for Two')
    plt.xlabel('Approximate Cost (₹)')
    plt.ylabel('Number of Restaurants')
    plt.tight_layout()
    plt.show()
```
